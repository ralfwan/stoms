# STOMS viewer in Jupyter

View ASE structures inline, adjust them, and save the view as an SVG (vector graphics). No ledger is needed. Install with `pip install "stoms[jupyter]"`; see `docs/jupyter.md`.

```python
from ase.build import bulk

from stoms.viewer import view

atoms = bulk("Si", "diamond", a=5.43)

viewer = view(atoms)
viewer
```

Drag to rotate, scroll to zoom. The toolbar toggles the cell and bonds, sets the atom size and the view direction; ⚙ has more settings, ⧉ opens the viewer in a new browser tab.

Settings from Python are shown at once:

```python
viewer.show_bonds = True
viewer.repeat = (2, 2, 2)
viewer.atom_scale = 0.9
viewer.set_view("a")
```

Save the view as an SVG **in a later cell** (the browser draws it; its drawing reaches Python when the cell that changed the view has finished):

```python
viewer.save("si.svg")
```

## Several structures and trajectories

A list, an ASE `Trajectory` or a file name: step through the frames with ◀ ▶, the slider, ← → or ▶ play.

```python
from ase.build import add_adsorbate, fcc111, molecule

images = []
for height in (3.0, 2.5, 2.0, 1.8):
    slab = fcc111("Pt", (3, 3, 3), vacuum=8.0)
    add_adsorbate(slab, molecule("CO"), height, "ontop")
    images.append(slab)

movie = view(images, settings={"bonds": True})
movie
```

```python
movie.frame = -1  # the last frame
movie.selected_atoms  # a copy of the shown frame, as ase.Atoms
```

## The same look for other structures

`get_settings()` holds the settings and the view (rotation, zoom); passed to `view()` it draws other structures the same way, for example several surface reconstructions with one cell.

```python
look = movie.get_settings()
view(images[0], settings=look)
```

## Outside the notebook

`save("x.html")` writes a standalone, interactive file; `open_in_browser()` opens it on the computer running the kernel (on a remote server, use ⧉ instead).

```python
movie.save("co-on-pt.html")
# movie.open_in_browser()
```
